Классы описывают реально существующие предметы и ситуации, а затем на основе этих описаний создаются объекты. При написании класса определяется общее поведенение для целой категории объектов.

Создаение объекта на основе класса называется созданием экземпляра. Т.е. мы работаем с экземплярами класса.



Функция являющаяся частью класса называется методом.
Метод __init__() - специальный метод, который автоматически выполняется при создании каждого нового экземпляра на базе класса.
self добавляется в определение, поскольку при будущем вызове метода __init__() (для создании экземпляра класса) Python автоматически передает аргумент self.
При каждом вызове метода, связанного с классом, автоматически передается self - ссылка на экземпляр; она предоставляет конкретному экземпляру доступ к атрибутам и методам класса

**Метод __init__() создает экземпляр, представляющий конкретную собаку, и присваивает его атрибутам name и age переданные значения. Затем Python возвращает экземпляр, представляющий собаку.
Он сохраняется в переменной my_dog

self.name - атрибут экземпляра или свойства объекта**

Атрибуты - это переменные, которые принадлежат классу или объекту. Они хранят данные.

self - ссылка на текущий объект(экзмепляр)

self – это способ для метода обратиться к самому объекту, для которого он был вызван. Без него невозможно сохранить состояние объекта.

```python
#dog.py

class Dog:
    """Простая модель собаки."""

    def __init__(self, name, age):#Метод создает экземпляр представляющий конкретную собаку, и присваивает его атрибутам name и age переданные значения
        """Инициализирует атрибиуты name и age.""" 
        self.name = name #переменная с префиксом self доступна для любого метода в классе.
        self.age = age#атрибут экземпляра - литчная переменная конкретного объекта
    
    def sit(self):
        """Имитирует, как собака садится по команде."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Имитирует, как собака перекатывается по команде."""
        print(f"{self.name} rolled over!")

#Создание экземпляра для конкретной собаки
my_dog = Dog('Willie', 6)
your_dog = Dog('Lucy', 3)
#После создания экземпляра на основе класса Dog можно применять точечную нотацию для вызова любых методов, определенных в Dog.
my_dog.sit()
my_dog.roll_over()

#Обращаемся к атрибутам экземпляра с помощью точенчной нотации
print(f"\nmy dog's name is {my_dog.name}.")
print(f"My dog is {my_dog.age} years old.")

print(f"\nYor dog's name is {your_dog.name}.")
print(f"Yor dog is {your_dog.age} years old.")
your_dog.sit()
```

```python
#9.1. Ресторан.

class Restaurant:
    def __init__(self, restaurant_name, cuisine_type):
        self.restaurant_name = restaurant_name
        self.cuisine_type = cuisine_type

    def describe_restaurant(self):
        """Выводит два атрибута экземпляра"""
        print(f"\nВ ресторане: {self.restaurant_name} вид кухни: {self.cuisine_type}.")

    def open_restaurant(self):
        """Выводит сообщение об открытии ресторана"""
        print(f"\nРесторан {self.restaurant_name} открыт с 10:00 до 22:00")

restautant = Restaurant('Pattison', 'italian')

print(f"Название ресторана: {restautant.restaurant_name}.")
print(f"Тип кухни: {restautant.cuisine_type}.")

restautant.describe_restaurant()
restautant.open_restaurant()
```

```python
#9.2. Три ресторана.

class Restaurant:
    def __init__(self, restaurant_name, cuisine_type):
        self.restaurant_name = restaurant_name
        self.cuisine_type = cuisine_type

    def describe_restaurant(self):
        """Выводит два атрибута экземпляра"""
        print(f"\nВ ресторане: {self.restaurant_name} вид кухни: {self.cuisine_type}.")

    def open_restaurant(self):
        """Выводит сообщение об открытии ресторана"""
        print(f"\nРесторан {self.restaurant_name} открыт с 10:00 до 22:00")

restautant_one = Restaurant('Pattison', 'Italian')
restautant_two = Restaurant('Pushkin Cafe', 'Russian')
restautant_three = Restaurant('Tanuki', 'Japan')

#Вызовем для каждого экземпляра метод descrier_restaurant
restautant_one.describe_restaurant()
restautant_two.describe_restaurant()
restautant_three.describe_restaurant()
```

```python
#9.3. Пользователи.

class User:
    def __init__(self, first_name, last_name, age, field, **more_info):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.field = field
        self.more_info = more_info

    def describe_user(self):
        """Выводит сводку с информацией о пользователе."""
        print(f"\nИнформация о {self.first_name} {self.last_name}")
        print(f"Место рождения {self.more_info.get('home_land')}, возраст {self.age}")
        print(f"Область занятий {self.field}.")

    def greet_user(self):
        """Выводит приветствие"""
        print(f"\nПривет {self.first_name} {self.last_name}!")

user_1 = User('Maks', 'Puzankov', 21, 'robotics', home_land = 'Russia')
user_2 = User('Stepan', 'Puzankov', 21, 'energetics', home_land = 'Russia')
user_3 = User('John', 'Stivenson', 35, 'space industry', home_land = 'USA')

user_1.describe_user()
user_1.greet_user()

user_2.describe_user()
user_2.greet_user()

user_3.describe_user()
user_3.greet_user()
```
