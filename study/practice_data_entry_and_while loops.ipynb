Циклы while
Цикл for получает коллекцию элементов и выполняет блок кода по одному разу
для каждого элемента в коллекции. В отличие от него, цикл while продолжает вы-
полняться, пока некое условие остается истинным.

```python
#counting.py

curent_number = 1

while curent_number <= 5:
    print(curent_number)
    curent_number += 1
```

```python
#parrot.py
#Программа может выполняться до тех пор, пока пользователь не захочет остановить ее. Т.е. в программе определяется признак
#завершения(quit value), и она работает, пока пользователь не введет нужное значение.

prompt = "\nTell me something, and I'll repeat it back to you:"
prompt += "\nEnter 'quit' to end the program."

message = ""

while message != 'quit':
    message = input(prompt)
    if message != 'quit':
        print(message)
```

```python
#parrot.py С добавлением флага
"""
Если программа должна выполняться только при истинности нескольких условий, то определите одну переменную — флаг
"""
prompt = "\nTell me something, and I'll repeat it back to you:"
prompt += "\nEnter 'quit' to end the program."

active = True #Изначально указываем True, чтобы программа начинала работу в активном состоянии.
while active: #Пока переменная active = True цикл выполняется
    message = input(prompt)

    if message == 'quit':
        active = False
    else:
        print(message)
```

```python
#cities.py
"""Оператор break и выход из цикла"""
prompt = "\nPlease enter the name of a city you have visited"
prompt += "\n(Enter 'quit' when you are finished)"

while True:
    city = input(prompt)
    if city =='quit':
        break
    else:
        print(f"I'd love to go to {city.title()}!")
```

Оператор break может использоваться в любых циклах Python. Например, его можно
добавить в цикл for для перебора элементов словаря.

```python
#counting.py
curent_number = 0

while curent_number < 10:
    curent_number += 1
    if curent_number % 2 == 0: #Если остаток равен 0, то оператор continue дает python указание проигнорировать оставшийся код
        continue               #цикла и вернуться к началу 

    print(curent_number)
```

Использование цикла while со списками и словарями
Чтобы работать с несколькими фрагментами информации, необходимо использовать в циклах while
списки и словари.
Цикл for хорошо подходит для перебора списков, но, скорее всего, список не должен
изменяться в цикле, поскольку у Python возникнут проблемы с отслеживанием
элементов списка. Изменять список в процессе обработки можно с помощью цикла
while. 

```python
#confirmed_users.py
#Перемещение элементов между списками

unconfirmed_users = ['alice', 'brian', 'candace']
confirmed_users = []

while unconfirmed_users:
    curent_user = unconfirmed_users.pop()

    print(f'Verifying user: {curent_user.title()}')
    confirmed_users.append(curent_user)

print("\nThe following users have been confirmed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
```

```python
#pets.py
#Удаление всех вхождений конкретного значения из списка

pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
print(pets)

del_pets = ['cat', 'dog']

for pet in del_pets:
    while pet in pets:
        pets.remove(pet)

print(pets)
```

```python
pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
del_pets = ['cat', 'dog']

pets = [pet for pet in pets if pet not in del_pets]

for pet in pets:
    print(pet.title())
```

```python
#mountain_poll.py
#Заполнение словаря данными, введенными пользователем

responses = {}
#Установка флага продолжения опроса
polling_active = True

while polling_active:
    #Запрос иимени и ответа пользователя.
    name = input("\nWhat is your name ?")
    response = input("Which mountain wold you like to climb someday ?")

    #Ответ сохраняется в словаре
    responses[name] = response

    #Проверка продолжения опроса
    repeat = input("Wold you like to let another person respond ? (yes/ no)")
    if repeat == 'no':
        polling_active = False

    #Опрос завершен, вывести результаты. Причем данный блок срабатывает именно тогда, когда
    #тело цикла завершится. 
print("\n--- Poll results")
for name, response in responses.items():
    print(f"{name.title()} would like to climb {response.title()}.")
```

```python
#7.8. Бутерброды

sandwich_orders = ['Submarine sandwich', 'Wrap', 'Club sandwich', 'Panini', 'Peanut butter']
finished_sandwiches = []

while sandwich_orders:
    sandwich = sandwich_orders.pop()
    print(f"I have prepared {sandwich.title()}")
    finished_sandwiches.append(sandwich)

print('\n--- Made sandwiches ---')
for finished_sandwich in finished_sandwiches:
    print(finished_sandwich.title())


```

```python
#7.9. Без пастромы
sandwich_orders = ['Submarine sandwich','pastrami', 'Wrap', 'Club sandwich','pastrami',
                    'Panini', 'Peanut butter', 'pastrami']

print('Unfortunately there is no more pastrami')

while 'pastrami' in sandwich_orders:
    sandwich_orders.remove('pastrami')

print('\n--- Ordered sandwiches ---')
for sandwich in sandwich_orders:
    print(sandwich.title())
```

```python
#7.10. Отпуск мечты

responses = {}

polling_active = True

while polling_active:
    name = input('What is your name ?')
    response = input('If you could visit one place in the world, where would you go ?')

    responses[name] = response

    repeat = input('Wold you like to let another person respond ? (yes/ no)')
    if repeat == 'no':
        polling_active = False

print('\n--- Poll results ---')
for name, response in responses.items():
    print(f"{name} would like to {response}")
    
```

```python
def make_album(artist, album):
    info_album = {}
    info_album['artist'] = artist; info_album['album'] = album
    return info_album

while True:
    print('\nTell me your favorite artist and his album')
    print("(enter 'q' at any time to quit)")

    artist_name = input('Введите имя исполнителя:')
    if artist_name == 'q':
        break
    album_name = input('Введите имя альбома: ') 
    if album_name == 'q':
        break

    template_album = make_album(artist_name, album_name)
    print(f"\n{template_album}")
```
