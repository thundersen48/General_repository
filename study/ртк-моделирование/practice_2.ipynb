```python
import sympy as sp

th1, th2, th3, th4, th5, th6 = sp.symbols('theta_1:7')
a2, a3 = sp.symbols('a_2 a_3')
d1, d4, d5, d6 = sp.symbols('d_1 d_4 d_5 d_6')

def dh_matrix(a, alpha, d, theta):
    c, s = sp.cos, sp.sin
    return sp.Matrix([
        [c(theta), -c(alpha)*s(theta),  s(alpha)*s(theta), a * c(theta)],
        [s(theta),  c(alpha)*c(theta), -s(alpha)*c(theta), a * s(theta)],
        [0,         s(alpha),           c(alpha),          d],
        [0,         0,                  0,                 1]
    ])

T1 = dh_matrix(0,  sp.pi / 2, d1, th1)
T2 = dh_matrix(-a2, 0,         0,  th2)
T3 = dh_matrix(-a3, 0,         0,  th3)
T4 = dh_matrix(0,  sp.pi / 2, d4, th4)
T5 = dh_matrix(0, -sp.pi / 2, d5, th5)
T6 = dh_matrix(0,  0,         d6, th6)

T06 = sp.trigsimp(T1 * T2 * T3 * T4 * T5 * T6)
#sp.pprint(T06)

# Численный расчет для заданной конфигурации углов
params = {
    th1: 0, th2: -sp.pi/2, th3: 0, th4: -sp.pi/2, th5: 0, th6: 0,
    d1: 0.181, a2: 0.613, a3: 0.571, d4: 0.135, d5: 0.120, d6: 0.117
}

sp.pprint(T06.subs(params))
```
