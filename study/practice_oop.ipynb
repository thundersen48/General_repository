Классы описывают реально существующие предметы и ситуации, а затем на основе этих описаний создаются объекты. При написании класса определяется общее поведенение для целой категории объектов.

Создаение объекта на основе класса называется созданием экземпляра. Т.е. мы работаем с экземплярами класса.



Функция являющаяся частью класса называется методом.
Метод __init__() - специальный метод, который автоматически выполняется при создании каждого нового экземпляра на базе класса.
self добавляется в определение, поскольку при будущем вызове метода __init__() (для создании экземпляра класса) Python автоматически передает аргумент self.
При каждом вызове метода, связанного с классом, автоматически передается self - ссылка на экземпляр; она предоставляет конкретному экземпляру доступ к атрибутам и методам класса

**Метод __init__() создает экземпляр, представляющий конкретную собаку, и присваивает его атрибутам name и age переданные значения. Затем Python возвращает экземпляр, представляющий собаку.
Он сохраняется в переменной my_dog

self.name - атрибут экземпляра или свойства объекта**

Атрибуты - это переменные, которые принадлежат классу или объекту. Они хранят данные.

self - ссылка на текущий объект(экзмепляр)

self – это способ для метода обратиться к самому объекту, для которого он был вызван. Без него невозможно сохранить состояние объекта.

```python
#dog.py

class Dog:
    """Простая модель собаки."""

    def __init__(self, name, age):#Метод создает экземпляр представляющий конкретную собаку, и присваивает его атрибутам name и age переданные значения
        """Инициализирует атрибиуты name и age.""" 
        self.name = name #переменная с префиксом self доступна для любого метода в классе.
        self.age = age#атрибут экземпляра - литчная переменная конкретного объекта
    
    def sit(self):
        """Имитирует, как собака садится по команде."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Имитирует, как собака перекатывается по команде."""
        print(f"{self.name} rolled over!")

#Создание экземпляра для конкретной собаки
my_dog = Dog('Willie', 6)
your_dog = Dog('Lucy', 3)
#После создания экземпляра на основе класса Dog можно применять точечную нотацию для вызова любых методов, определенных в Dog.
my_dog.sit()
my_dog.roll_over()

#Обращаемся к атрибутам экземпляра с помощью точенчной нотации
print(f"\nmy dog's name is {my_dog.name}.")
print(f"My dog is {my_dog.age} years old.")

print(f"\nYor dog's name is {your_dog.name}.")
print(f"Yor dog is {your_dog.age} years old.")
your_dog.sit()
```

**Наследование**
Если класс представляет собой специализированную версию ранее написанного класса, то можно воспользоваться наследованием.
Один класс, наследующий от другого, автоматически получает все атрибуты и методы первого класса. Исходный класс называется родителем, а новый класс - потомком. Класс потомок наследует атрибуты и методы класса-родителя, но при этом мождет определять и собственные атрибуты и методы 

super() - специальная функция, которая позволяет вызывать метод родительского класса. Благодаря чему Python получает указание вызвать метод __init__() класса Car, в результате чего экземпляр ElectricCar имеет доступ ко всем атрибутам класса родителя.
Имя super проихсодит из общепринятой терминологии: класс-родитель называтся суперклассом, а класс-потомок - подклассом.

super().__init__(...) вызывает конструктор (__init__) родительского класса, передавая ему аргументы, и применяет этот код к текущему объекту (который является экземпляром дочернего класса).
В результате текущий объект получает все атрибуты, определённые в родительском конструкторе, как свои собственные.

```python
#electic_car.py
class Car:
    """Простая модель автомобиля."""
    def __init__(self, make, model, year):
        """Инициализирует атрибуты описания автомобиля."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0 #Назначили атрибуту значение по умолчанию

    def get_descriptive_name(self):
        """Возвращает отформатированное описание."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Выводит данные о пробеге машины в километрах."""
        """Новый метод read_odometer, упрощает чтение данных о пробеге машины в км. """
        print(f"This car has {self.odometer_reading} km on it.")

    def update_odometer(self, mileage):
        """Устанавливает заданное значение на одометре.
        При попытке обратной прокрутки изменение отклоняется.
        """
        if mileage >= self.odometer_reading:#Если новое значение пробега больше или равно ему, то показания  
            self.odometer_reading = mileage#одометра обновляются с помощью нового значения
        else:
            print("Вы не можете подкрутить назад значения одометра!")#Если новое значение km меньше текущего, то выскочит предупрежедение

    def increment_odometer(self, km):
        """Увеличивает показания одометра с заданным приращением."""
        current_odometer = self.odometer_reading
        new_odometer = km + self.odometer_reading
        if new_odometer >= current_odometer:
            self.odometer_reading = new_odometer
        else:
            print("Вы не можете подкрутить назад значения одометра!")


class ElectricCar(Car):
    """Предоставляет аспекты машины, спецефические для электромобилей."""
    def __init__(self, make, model, year): #Метод получает информацию, необходимую для создания экземпляра Car.
        """Инициализирует атрибуты класса-родителя"""
        super().__init__(make, model, year)#Вызываем метод родительского класса, чтобы получить доступ ко всем атрибутам класса-родителя.

my_leaf = ElectricCar('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
```

Любой метод родительского класса, который делает не то что нужно, можно переопределить. Для этого в классе потомке определяется метод с тем же именем, что и у метода родительского класса. Python игнорирует метод родителя и обращает внимание только на метод, определенный в классе потомке.

Композиция - это отношение часть-целое, при котором часть не может существовать или использоваться отдельно от целого.

**Явные примеры композиции**
Композиция — это подход в ООП, при котором один класс управляет объектами других классов через отношение «содержит / имеет» (has-a), а не через наследование «является» (is-a).

```python
#Автомобиль и двигатель

class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f"Двигатель ({self.horsepower} л.с.) запущен."

class Car:
    def __init__(self, model, horsepower=0):
        self.model = model
        self.engine = Engine(horsepower)

    def drive(self):
        print(self.engine.start())
        print(f"{self.model.title()} поехала.")

my_car = Car("porshe 911", 450)
my_car.engine.start()
my_car.drive()
```

```python
#Корзина интернет-магазина

class OrderItem:#Товар в строчке чека
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def get_total(self):
        return self.price * self.quantity

class Order:
    def __init__(self):
        self.items = []#Композиция - экземпляр Order содержит в себе список объектов другого класса.

    def add_item(self, item):
        self.items.append(item)

    def calculate_total(self):
        return sum(item.get_total() for item in self.items)


order = Order()
# Создаем объект OrderItem и сразу отдаем его внутри Order
order.add_item(OrderItem("Механическая клавиатура", 8000.0, 1))
order.add_item(OrderItem("Монитор", 25000.0, 2))

print(f"Итого к оплате: {order.calculate_total()} руб.")  

```

```python
#Композиция - это когда класс содержит внутри себя экземпляр другого класса в качестве своего атрибута
#PizzaShop.py

class Employee:
    def __init__(self, name, salary=0):
        self.name = name
        self.salary = salary

    def work(self):
        print(self.name, "делает общую работу")

class Chef(Employee):
    def __init__(self, name):
        super().__init__(name, 5000)

    def work(self):
        print(f"Повар {self.name} готовит пиццу.")

class Server(Employee):
    def __init__(self, name):
        super().__init__(name, 4000)

    def work(self):
        print(f"Официант {self.name} обслуживает клиента.")

class PizzaShop:
    def __init__(self):
        # КОМПОЗИЦИЯ: Пиццерия создает и внутри себя хранит объекты работников
        self.server = Server('Павел')
        self.chef = Chef('Борис')

    def order(self, customer_name):
        print(f"Принят заказ от {customer_name}")
        self.server.work() # Делегирование задачи официанту
        self.chef.work() # Делегирование задачи повару

shop = PizzaShop()
shop.order("Алексей")
```

```python
#«Компьютер и Процессор» (Внутреннее владение)

class Processor:
    def __init__(self, model, cores):
        self.model = model
        self.cores = cores

    def process(self, task):
        return f"Процессор {self.model} выполняет: {task}"

class Computer:
    def __init__(self, brand, model, cores):
        self.brand = brand
        self.processor = Processor(model, cores)

    def run_app(self, app_name):
        print(self.processor.process(app_name))

pc = Computer("ASUS", "Intel i7", 8)
pc.run_app("Photoshop")
```

```python
#«Персонаж и Оружие» (Гибкая замена компонента)
class Sword:
    def __init__(self):
        pass

    def use(self):
        """Возвращает результат удара мечом."""
        print(f"Удар мечом на 20 урона!")

class Bow:
    def __init__(self):
        pass

    def use(self):
        """Возвращает результат урона из лука."""
        print("Выстрел из лука на 12 урона!")

class Hero:
    def __init__(self, name, weapon):
        self.name = name
        self.weapon = weapon#Принимает объект оружия при создании

    def attack(self):
        """Вызываем метод use() у текущего оружия и печатает результат."""
        self.weapon.use()

    def change_weapon(self, new_weapon):
        """Позволяет изменить оружие персонажа."""
        self.weapon = new_weapon

hero = Hero("Артур", Sword())
hero.attack()

hero.change_weapon(Bow())
hero.attack()
```

```python
class Sensor:
    def __init__(self, name, value):
        self.name = name
        self.value = value

    def read(self):
        return f"{self.name}: {self.value}"

class SmartHome:
    def __init__(self):
        self.sensors = []

    def add_sensor(self, sensor):
        self.sensors.append(sensor)

    def show_status(self):
        for sensor in self.sensors:
            print(sensor.read())

    def get_average_reading(self):
        average = sum(sensor.value for sensor in self.sensors) / len(self.sensors)
        return average

home = SmartHome()
home.add_sensor(Sensor("Термометр", 22.5))
home.add_sensor(Sensor("Датчик влажности", 45.0))

home.show_status()
print("Среднее значение:", home.get_average_reading())
```

```python
#order_system.py

class EmailSender:
    def __init__(self):
        pass

    def send(self, recipient, text):
        return f"[EMAIL на {recipient}]: {text}"

class SmsSender:
    def __init__(self):
        pass

    def send(self, recipient, text):
        return f"[SMS на {recipient}]: {text}"

class NotificationService:
    def __init__(self, sender):
        self.sender = sender

    def send_order_confirmation(self, user_contact, order_id):
        text = f"Заказ №{order_id} успешно оформлен!"
        return self.sender.send(user_contact, text)#опять вызваем return, потому что он не передается автоматически наверх по
                                                   #цепочке вызовов.
        

    def set_sender(self, new_sender):
        self.sender = new_sender


service = NotificationService(EmailSender())
print(service.send_order_confirmation("user@mail.com", 1001))

service.set_sender(SmsSender())
print(service.send_order_confirmation("+79991112233", 1002))
    
```

```python
#subscription_system.py

class YoomoneyGateway:

    def pay(self, amount, account):
        return f"[ЮMoney]: Списано {amount} руб. со счета {account}"

class TelegramPayGateway:
    def pay(self, amount, account):
        return f"[Telegram Pay]: Списано {amount} руб. с аккаунта {account}"

class ApplePay:
    def pay(self, amount, account):
        return f"[Apple Pay]: Списано {amount} руб. с аккаунта {account}"

class Payment_Processor:
    def __init__(self, gateway):
        self.gateway = gateway#Композиция. При инициализации он принимаем объект платежного шлюза gateway и сохраяет его.

    def process_subscription(self, user_account, price):
        return self.gateway.pay(price, user_account)

    def change_gateway(self, new_gateway):
        """Позволяет сменить шлюз оплаты"""
        self.gateway = new_gateway

processor = Payment_Processor(YoomoneyGateway())
processor.process_subscription("thundersen48", 9100)

processor.change_gateway(ApplePay())
processor.process_subscription("thundersen48", 500)
```

```python
#logging_system.py

class HTTPRequest:
    """
    url- адрес страницы
    status_code - HTTP-код ответа
    response_size - размер ответа в байтах
    """
    def __init__(self, url:str, status_code:int, response_size:int):
        self.url = url
        self.status_code = status_code
        self.response_size = response_size

class TrafficAnalyzer:
    def __init__(self):
        self.request = []#Empty list for save to response

    def add_request(self, request):
        """Принимает объект класса HTTPRequest"""
        self.request.append(request)
        

    def get_total_bytes(self):
        """Возвращает суммарный объем переданных байт 
        по всем сохраненным запросам."""
        return sum(getattr(req, 'response_size', 0) for req in self.request)
        #return sum(req.response_size for req in self.request if isinstance(req.response_size, (int, float)))

    def get_failed_requests(self):
        """Возвращает список объектов у которых запросы с ошибкой."""
        failed = []
        for req in self.request:#Перебираем все объекты
            # Берем статус-код у текущего объекта и сравниваем с 400
            if req.status_code >= 400:
                failed.append(req)

        return failed            

analyzer = TrafficAnalyzer()
analyzer.add_request(HTTPRequest("/home", 200, 1500))
analyzer.add_request(HTTPRequest("/login", 404, 500))
analyzer.add_request(HTTPRequest("/dashboard", 200, 3000))

print("Всего байт:", analyzer.get_total_bytes())

failed = analyzer.get_failed_requests()
print("Ошибок найдено:", len(failed))

print("URL ошибки:", failed[0].url)#Обращаеся к объекту с 0 индексом в списке failed и вызываем у него атрибут url.
```

```python
#Архитектура игрового персонажа (Сборный экипировочный юнит)
#В RPG-игре создается механика персонажей. Вся защита персонажа складывается из базовой защиты самого героя и суммы брони всех 
#надетых на него предметов экипировки."""

class ArmorItem:
    """
    name - название предмета
    defense - показатель защиты
    """
    def __init__(self, name:str, defense:int):
        self.name = name
        self.defense = defense

class Weapons:
    def __init__(self, name:str, damage:int):
        self.name = name
        self.damage = damage

class Character:
    """
    name - имя персонажа
    base_defense - базовая защита без экипировки
    equipment - список для хранения объектов элементов экипировки
    """
    def __init__(self, name:str, base_defense:int, base_damage=3):
        self.name = name
        self.base_defense = base_defense
        self.base_damage = base_damage
        self.equipment = []#composition
        self.weapons = []

    def equip(self, item):
        self.equipment.append(item)

    def add_weapon(self, weapon):
        self.weapons.append(weapon)

    def get_total_defense(self):
        sum_equip = sum(equip.defense for equip in self.equipment) + self.base_defense
        return sum_equip

    def get_total_damage(self):
        sum_damage = sum(weap.damage for weap in self.weapons) + self.base_damage 
        return sum_damage

    def get_equipment_list(self):
        """возвращает список строк с названиями всех надетых предметов"""  
        return [equip.name for equip in self.equipment]; [weap.name for weap in self.weapons]
    

hero = Character("Рыцарь", base_defense=10)

hero.equip(ArmorItem("Стальной шлем", 5))
hero.equip(ArmorItem("Кольчуга", 15))

hero.add_weapon(Weapons("Стальной меч", 12))
print("Общая защита:", hero.get_total_defense())
print("Общий урон:", hero.get_total_damage())
print("броня:", hero.get_equipment_list())

```

```python
#Архитектура игрового персонажа (Сборный экипировочный юнит)
#В RPG-игре создается механика персонажей. Вся защита персонажа складывается из базовой защиты самого героя и суммы брони всех 
#надетых на него предметов экипировки."""

class ArmorItem:
    """
    name - название предмета
    defense - показатель защиты
    """
    def __init__(self, name:str, defense:int):
        self.name = name
        self.defense = defense

class Weapons:
    def __init__(self, name:str, damage:int):
        self.name = name
        self.damage = damage

class Character:
    """
    name - имя персонажа
    base_defense - базовая защита без экипировки
    equipment - список для хранения объектов элементов экипировки
    """
    def __init__(self, name:str, base_defense:int, base_damage=3):
        self.name = name
        self.base_defense = base_defense
        self.base_damage = base_damage
        self.equipment = []#composition

    def equip(self, item):
        self.equipment.append(item)

    def add_weapon(self, weapon):
        self.equipment.append(weapon)

    def get_total_defense(self):
        """getattr(equip, 'defense', 0) - Эта функция пробует взять у объекта equip атрибут defense.
          Если атрибут есть (как у ArmorItem) – возвращает его значение.
          Если атрибута нет (как у Weapons) – возвращает значение по умолчанию, которое я указал как 0.
          """
                
        sum_equip = sum(getattr(equip, 'defense', 0) for equip in self.equipment) + self.base_defense
        return sum_equip

    def get_total_damage(self):
        sum_damage = sum(getattr(weap, 'damage', 0) for weap in self.equipment) + self.base_damage 
        return sum_damage

    def get_equipment_list(self):
        """возвращает список строк с названиями всех надетых предметов"""  
        return [equip.name for equip in self.equipment]
    

hero = Character("Рыцарь", base_defense=10)

hero.equip(ArmorItem("Стальной шлем", 5))
hero.equip(ArmorItem("Кольчуга", 15))

hero.add_weapon(Weapons("Стальной меч", 12))
print("Общая защита:", hero.get_total_defense())
print("Общий урон:", hero.get_total_damage())
print("Снаряжение:", hero.get_equipment_list())

```
