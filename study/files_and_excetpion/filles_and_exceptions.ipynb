Исключения(exceptions) - специальные объекты, которые создаются для управления ошибками, возникающими во время выполнения программ Python.

```python
"""1. Чтение всего файла"""
#file_reader.py
# Объект Path, указывающий на файл, позволяет вам выполнить многие действия.
# Например, прежде чем работать с файлом,вы можете проверить, существует ли он, 
# прочитать его содержимое или записать в него новые данные.

from pathlib import Path

path = Path('pi_digits.txt')# Создаем объект Path, представляющий собой файл pi_digits
contents = path.read_text()#метод read_text() читает все содержимое файла и сохраняет это содержимое в одной длинной строке в переменной contents.
contents = contents.rstrip()#Удаляет пустую строку, которая возвращается методом read_text() при чтении, если достигнут конец файла.
                            #когда достигнут конец файлав конце вывдо
print(contents)

#Можно удалить символ новой строки при чтении содержимово файла, применив метод rstrip() после метода read_text().
contents = path.read_text().rstrip()#Python применяет метод read_text(), затем интерпретатор применяет метод rstrip() к строке возвращаемой
#методом read_text(). Итоговая строка присваивается переменной contents. Такой подход называется цепочкой методов (method chaining) и используется довольно-таки часто
```
