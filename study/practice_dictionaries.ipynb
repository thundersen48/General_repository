Словари - это структура данных, предназначенная для объединения взаимосвязанной информации.

Словарь в языке Python представляет собой совокупность пар «ключ — значение».
Каждый ключ связывается с неким значением, и программа может получить зна-
чение, связанное с заданным ключом. Значением могут быть число, строка, список
и даже другой словарь. Собственно, любой объект, создаваемый в программе Python,
может стать значением в словаре.

```python
#Отслеживание позиции пришельца, который может двигаться с разными скоростями

alien_0 = {'x_position': 0, 'y_position': 25, 'speed': 'fast'}

print(f'Current position: {alien_0["x_position"]}; {alien_0["y_position"]}')

if alien_0['speed'] == 'slow':
    x_increment = 1
elif alien_0['speed'] == 'medium':
    x_increment = 2
else:
    x_increment = 3

#Новая позиция равна сумме старой позиции и приращения.
alien_0['x_position'] = alien_0['x_position'] + x_increment

print(f'New position: {alien_0["x_position"]}; {alien_0["y_position"]}')
```

Вложение(nesting) - сохранение множества словарей в списке или сохранение списка как значение элемента словаря

```python
#aliens.py

"""Три словаря, каждый из которых представляет отдельного пришельца"""

alien_0 = {'color': 'green', 'points': 5}
alien_1 = {'color': 'yellow', 'points': 10}
alien_2 = {'color': 'red', 'points': 15}

aliens = [alien_0, alien_1, alien_2]

for alien in aliens:
    print(alien)
```

```python
#Создаем пустой список для хранения пришельцев
aliens = []

for alien_number in range(30):
    new_alien = {'color': 'green', 'points': 5, 'speed': 'slow'}
    aliens.append(new_alien)

#Вывод данных о первых пяти пришельцах
for alien in aliens[:5]:
    print(alien)
print("...")

#Вывод количества созданных пришельцев
print(f'Total number of aliens: {len(aliens)}')
```

```python
#Создаем пустой список для хранения пришельцев
aliens = []

for alien_number in range(20):
    new_alien = {'color': 'green', 'points': 5, 'speed': 'slow'}
    aliens.append(new_alien)

for alien_number in range(10):
    new_alien = {'color': 'yellow', 'points': 10, 'speed': 'medium'}
    aliens.append(new_alien)
#Если пришелец зеленый, то его цвет меняется на желтый ('yellow'), скорость на среднюю ('medium'), 
#а награда увеличивается до 10 очков
for alien in aliens[0:]:
    if alien['color'] == 'green':
        alien['color'] = 'yellow'
        alien['speed'] = 'medium'
        alien['points'] = 10
    elif alien['color'] == 'yellow':
        alien['color'] = 'red'
        alien['speed'] = 'fast'
        alien['points'] = 15

#Вывод данных о первых пяти пришельцах
for alien in aliens[10:]:
    print(alien)
print("...")

#Вывод количества созданных пришельцев
print(f'Total number of aliens: {len(aliens)}')
```

```python
aliens = []

for alien_number in range(30):
    new_alien = {'color': 'green', 'points': 5, 'speed': 'slow'}
    aliens.append(new_alien)

for index, alien in enumerate(aliens[0:]):
    if index < 5:
        alien['color'] = 'yellow'
        alien['speed'] = 'medium'
        alien['points'] = 10
    elif index >= 5 and index < 20:
        alien['color'] = 'red'
        alien['speed'] = 'fast'
        alien['points'] = 15
    else:
        pass

for alien in aliens:
    print(alien)
```

```python
#favorite_languages.py

favorite_languages = {
    'jen': ['python', 'rust'],
    'sarah': ['c'],
    'edward': ['rust', 'go'],
    'phil': ['python', 'haskell'],
}
#При переборе словаря переменная languages используется для хранения каждого значения из него, поскольку иы знаем, что оно будет
# представлять собой список. 
for index, (name, languages) in enumerate(favorite_languages.items(),start=1):
    if len(languages) >= 2:
        print(f'\n{index}. {name.title()}\'s favorite languages are:')
        for language in languages:
            print(f'\t{language.title()}')
    elif len(languages) == 1:
        print(f'\n{index}. {name.title()}\'s favorite language is {languages[0].title()}')
```
