```python
#greeter.py

def greet_user(): #опеределение функции
    """Выводит приветствие."""#Строка документации, которая представляет описание действий функции(Строк может быть несколько)
    print('Hello!') #Тело функции 

greet_user()#Вызов функции путем указания ее имени.
```

В данном случае аргумент 'Maks' был передан функции greet_user(), а его значение было сохранено в переменной username

```python
#printing_models.py

unprinted_designs = ['phone case', 'robot pedant', 'dodecahedron']
completed_models = []

while unprinted_designs:
    current_design = unprinted_designs.pop()
    print(f"Printing model: {current_design}")
    completed_models.append(current_design)

print("\nThe following models have been printed:")
for completed_model in completed_models:
    print(completed_model) 
```

```python
#Изменим структуру кода, добавив две функции, каждая из которых решает однку конкретную задачу.
unprinted_designs = ['phone case', 'robot pedant', 'dodecahedron']
completed_models = []

def print_models(unprinted_designs, completed_models):
    """
    Имитирует печать модели, пока список не станет пустым.
    Каждая модель после печати перемещается в completed_models
    """
    while unprinted_designs:
        current_design = unprinted_designs.pop()
        print(f"Printing model: {current_design}")
        completed_models.append(current_design)

def show_completed_models(completed_models):
    """Выводит информацию обо всех напечатанных моделях"""
    print("\nThe following models have been printed:")
    for completed_model in completed_models:
        print(completed_model)

print_models(unprinted_designs[:], completed_models)
show_completed_models(completed_models)
```

```python
#8.10. Отправка сообщений

messages = ['hello', 'hi', 'good evening!']
sent_messages = []

def send_messages(msgs, send_msgs):
    """Выводит каждое сообщение и перемещает его в список sent_messages"""
    print(f"Вывод сообщений:\n")
    while msgs:
        msg = msgs.pop()
        print(msg)
        send_msgs.append(msg)

send_messages(messages, sent_messages)

print(f"\nсписок сформированных сообщений - {messages}")
print(f"список отправленных сообщений - {sent_messages}")
```

**Передача произвольного набора элементов**

В некоторых случая неизвестно сколько аргументов должно быть передано функции. Python позволяет функции получать произвольное кол-во аргументов из вызывающего оператора.

```python
#pizza.py

def make_pizza(*toppings):#параметр для получения произвольного кол-ва аргументов
    """Выводит описание пиццы."""
    print(f"\nMaking a pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")

make_pizza('pepperoni')
make_pizza('mushrooms', 'green peppers', 'extra cheese')

#Благодрая звездочке python получает указание создать пустой кортеж toppings и упаковать в него все полученны значения.
#Данный синтаксис работает независимо от кол-ва аргументов, переданных функции
```

**Позиционные аргументы с произвольными наборами аргументов**
Чтобы функция могла вызываться с разным кол-ом аргументов, то параметр для получения произвольного кол-ва аргументов должен стоять на последнем месте в определении функции.
Python сначала подбирает соответствия для позиционных и именованных аргументов, а затем объединяет все остальные аргументы в последнем параметре

```python
#Например, если функция должна получать размер пиццы, то данный параметр должен стоять в списке до парамета *toppings

def make_pizza(size, *toppings):
    """выводит описание пиццы."""
    print(f"\nMaking a {size}-cm pizza with the folowing toppings:")
    for topping in toppings:
        print(f"- {topping}")

make_pizza(24, 'peperoni')
make_pizza(22, 'mushrooms', 'green peppers', 'extra cheese')
```

**Примечание**
Часто используется имя обобщенного параметра *args, который служит для храения произвольного набора позиционных аргументов.

**Использование произвольного набора именнованых аргументов**

Две звездочки в параметре функции заставляют Python создать пустой словарь и упаковать в него все пары "имя-значение".

```python
#user_profile.py

#Определение build_profile() ожидает получить имя и фамилию, а также позволяет передать любое кол-во пар "имя-значение".
def build_profile(first, last, **user_info):
    """Создает словарь, содержащий информацию о пользователе."""
    user_info['first_name'] = first#Внутри функции можно обращаться к парам "имя-значение" из user_info как в любом словаре.
    user_info['last_name'] = last
    return user_info #Возвращаем словарь user_info в точку вызова функции.

user_profile = build_profile('albert', 'einstein', location = 'princeton', field = 'physics')
print(user_profile)
```

**Примечание**
В программах часто используется имя обобщенного параметра **kwargs, который служит для хранения проивзольного набора ключевых аргументов.

```python
#8.12. Бутерброды.

sandwich_сomponents = ['cheese','bread', 'pepper', 'tomatoes', 'oil', 'meat']
sandwich_сomponents_two = ['bread', 'fish', 'mayonnaise', 'cucumbers']
sandwich_сomponents_three = ['ketchup', 'sausage', 'cheese']

def sandwich(components, number):
    print(f"\nСостав {number} заказанного бутерброда:")
    for component in components:
        print(component)

sandwich(sandwich_сomponents, 1)
sandwich(sandwich_сomponents_two, 2)
sandwich(sandwich_сomponents_three, 3)
```
